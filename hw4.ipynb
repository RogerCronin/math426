**Exercise 3.2.4**

Prove that if $A$ is an invertible square matrix, then $A^+=A^{-1}$

$$A^+=(A^TA)^{-1}A^T=(A^{-1}(A^T)^{-1})A^T=A^{-1}((A^T)^{-1}A^T)=A^{-1}I=A^{-1}$$

**Exercise 3.3.4**

Prove that if the $m\times n,(m\geq n)$ matrix $A$ is not rank-deficient, then the factor $\hat{R}$ of the thin QR factorization is invertible. (Hint: Suppose on the contrary that $\hat{R}$ is singular. Show using the factored form of $A$ that this would imply that $A$ is rank-deficient.)

Given some $A=\hat{Q}\hat{R}$ where $\hat{R}$ isn't invertible, WTS $A$ isn't invertible.

Since $\hat{R}$ isn't invertible, $\det(\hat{R})=0$. Then $\det(A)=\det(\hat{Q}\hat{R})=\det(\hat{Q})\cdot0=0$ means A isn't invertible.

**Exercise 3.3.8**

The matrix $P=\hat{Q}\hat{Q}^T$ derived from the thin QR factorization has some interesting and important properties.

(a) Prove that $P=AA^+$.

For sake of notation, let $\hat{Q}=Q$ and $\hat{R}=R$.

$$AA^+=(QR)((QR)^T(QR))^{-1}(QR)^T=
(QR)(R^{-1}Q^{-1})(Q^T)^{-1}(R^T)^{-1}(R^TQ^T)=
Q(RR^{-1})(Q^TQ)^{-1}((R^T)^{-1}R^T)Q^T=
QII^{-1}IQ^T=QQ^T$$

(b) Prove that $P^2=P$. (This property defines a *projection matrix*.)

$$P^2=(QQ^T)(QQ^T)=Q(Q^TQ)Q^T=QIQ^T=QQ^T=P$$

(c) Any vector $x$ may be written trivially as $x=u+v$, where $u=Px$ and $v=(I-P)x$. Prove that $u$ and $v$ are orthogonal. (Together with part (b), this means that $P$ is an *orthogonal projector*.)

First, know that $P^T=(QQ^T)^T=QQ^T=P$.

WTS $u^Tv=0$.

$$u^Tv=(Px)^T(I-P)x=(x^TP^T-x^TP^TP)x=(x^TP-x^TP)x=0$$

Thus $u$ and $v$ are orthogonal.

**Exercise 4.1.1**

$x^2=e^{-x}$, over $[-2,2]$

$f(x)=x^2-e^{-x}=0$

```python
def f(x):
    return x ** 2 - np.exp(-x)

x = np.linspace(-2, 2, 100)
y = np.vectorize(f)(x)

plt.axhline(0, linestyle = "--", color = "r")
plt.plot(x, y)
```

Looks to be 1 root around $x=0.7$.

```python
print(root_scalar(f, bracket = [0.5, 1]).root)
```

$\kappa_r=|2r+e^{-r}|^{-1}$

$\kappa_{0.703}\approx0.526$

**Exercise 4.1.2**

$2x=\tan(x)$, over $[-0.2,1.4]$

$f(x)=2x-\tan(x)=0$

```python
def f(x):
    return 2 * x - np.tan(x)

x = np.linspace(-0.2, 1.4, 100)
y = np.vectorize(f)(x)

plt.axhline(0, linestyle = "--", color = "r")
plt.plot(x, y)
```

Looks to be 2 roots around $x=0, 1.2$.

```python
print(root_scalar(f, bracket = [-0.5, 0.5]).root)
print(root_scalar(f, bracket = [0.8, 1.2]).root)
```

$\kappa_r=|2-\sec^2(x)|^{-1}$

$\kappa_{0}=1$, $\kappa_{1.166}\approx0.007$
