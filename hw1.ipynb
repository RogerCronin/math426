**Exercise 1.1.2**

Prove that (1.1.5) is equivalent to (1.1.6). This means showing first that (1.1.5) implies (1.1.6), and then separately that (1.1.6) implies (1.1.5).

(1.1.5) states $\frac{|\text{fl}(x)-x|}{|x|}\leq\frac{2^{n-d-1}}{2^n}=\frac{1}{2}\epsilon_M$, and (1.1.6) states $\text{fl}(x)=x(1+\epsilon)$ for some $|\epsilon|\leq\frac{1}{2}\epsilon_M$.

(1.1.5) $\to$ (1.1.6):

$$\begin{align*}
&\frac{|\text{fl}(x)-x|}{|x|}=|\frac{\text{fl}(x)-x}{x}|\leq\frac{1}{2}\epsilon_M\\
\to&\,-\frac{1}{2}\epsilon_M\leq\frac{\text{fl}(x)-x}{x}\leq\frac{1}{2}\epsilon_M\\
\end{align*}$$

Now assume $x>0$.

$$\begin{align*}
\to&\,\text{fl}(x)-x\leq x\frac{1}{2}\epsilon_M\\
\to&\,\text{fl}(x)\leq x(1+\frac{1}{2}\epsilon_M)
\end{align*}$$

Now if we set $\text{fl}(x)=x(1+\epsilon)$ where $\epsilon\leq\frac{1}{2}\epsilon_M$, we get exactly (1.1.6) for positive $\epsilon$.

If we instead assume $x<0$,

$$\begin{align*}
\to&\,-x\frac{1}{2}\epsilon_M\geq\text{fl}(x)-x\\
\to&\,x(1-\frac{1}{2}\epsilon_M)\geq\text{fl}(x)\\
\end{align*}$$

Now if we set $\text{fl}(x)=x(1+\epsilon)$ where $-\epsilon\leq\frac{1}{2}\epsilon_M$, we get exactly (1.1.6) for negative $\epsilon$.

We can combine these into the given $|\epsilon|\leq\frac{1}{2}\epsilon_M$.

(1.1.6) $\to$ (1.1.5):

$$\begin{align*}
&\text{fl}(x)=x(1+\epsilon),|\epsilon|\leq\frac{1}{2}\epsilon_M\\
\to&\,-\frac{1}{2}\epsilon_M\leq\epsilon\leq\frac{1}{2}\epsilon_M\\
\end{align*}$$

Now assume positive $\epsilon$.

$$\to\,\text{fl}(x)\leq x(1+\frac{1}{2}\epsilon_M)$$

We can walk back the algebraic manipulation shown in (1.1.5) $\to$ (1.1.6) to reach $\frac{|\text{fl}(x)-x|}{|x|}\leq\frac{1}{2}\epsilon_M$.

If we instead assume negative $\epsilon$,

$$\to\,x(1-\frac{1}{2}\epsilon_M)\geq\text{fl}(x)$$

We can again walk back the algebraic manipulation to reach the same $\frac{|\text{fl}(x)-x|}{|x|}\leq\frac{1}{2}\epsilon_M$.



As (1.1.5) $\to$ (1.1.6) and (1.1.6) $\to$ (1.1.5), they are equivalent.

Generalize (1.2.13) to finding a root of the $n$&ZeroWidthSpace;th degree polynomial $p(x)=a_nx^n+\cdots+a_1x+a_0$, and show that the relative condition number of a root $r$ with respect to perturbations only in $a_k$ is $\kappa_r(a_k)=|\frac{a_kr^{k-1}}{p'(r)}|$.

(1.2.13) states:

$$\kappa_f(a)=|\frac{a}{r_1}\cdot\frac{dr_1}{da}|=|\frac{ar_1}{2ar_1+b}|=|\frac{r_1}{r_1-r_2}|$$

For finding the conditioning of the leading coefficient of some quadratic $at^2+bt+c=0$ on some root $f(a)=r_1$.

For a general $n$&ZeroWidthSpace;th degree polynomial of the form $p(x)=a_nx^n+\cdots+a_1x+a_0$, we can follow the same steps outlined in the textbook.

Differentiating $a_nr_1^n+\cdots+a_kr_1^k+\cdots+a_1r_1+a_0$ implicitly with respect to $a_k$, holding all other $a_i$ fixed, yields

$$\begin{align*}
&\frac{d}{da_k}(a_nr_1^n+\cdots+a_kr_1^k+\cdots+a_1r_1+a_0)\\
=\,& a_n(n)r_1^{n-1}\frac{dr_1}{da_k}+\cdots+(r_1^k + a_k(k)r_1^{k-1}\frac{dr_1}{da_k})+\cdots+a_1(1)r_1^0\frac{dr_1}{da_k}\\
=\,& \frac{dr_1}{da_k}(a_n(n)r_1^{n-1}+\cdots+a_k(k)r_1^{k-1}+\cdots+a_1)+r_1^k\stackrel{\text{SET}}{=}0\\
\to\,& \frac{dr_1}{da_k}=\frac{-r_1^k}{a_n(n)r_1^{n-1}+\cdots+a_1}
\end{align*}$$

Now that we have a generalized equation for $\frac{dr_1}{da_k}$ and indeed any $\frac{dr}{da_k}$, we can find the condition number of a root $r$ with respect to perturbations only in $a_k$.

$$\kappa_r(a_k)=|\frac{a_k(dr/da_k)}{r}|=|\frac{a_k}{r}\cdot\frac{-r_1^k}{a_n(n)r_1^{n-1}+\cdots+a_1}|=|\frac{a_kr^{k-1}}{p'(r)}|$$

**Exercise (1.4.2)**

Let $f(x)=\frac{e^x-1}{x}$.

(a) Find the condition number $\kappa_f(x)$. What is the maximum of $\kappa_f(x)$ over $-1\leq x\leq1$?

As calculated in 1.2.3b,

$$\kappa_f(x)=|\frac{e^x(x-1)+1}{e^x-1}|$$

By way of Desmos, the maximum of $\kappa_f(x)$ over $-1\leq x\leq1$ is at the right bound $\kappa_f(1)=\frac{1}{e-1}$.

(b) Use the "obvious" algorithm `(exp(x) - 1) / x` to compute $f(x)$ at $x=10^{-2},10^{-3},10^{-4},\dots,10^{-11}$.

```python
import numpy as np

x_values = [10 ** e for e in range(-2, -11 - 1, -1)]

def obvious(x):
    return (np.exp(x) - 1) / x

obvious_algorithm = []
for x in x_values:
    y = obvious(x)
    obvious_algorithm.append(y)
    print(f"f({x:e}) = {y}")
```

(c) Create a second algorithm from the first 8 terms of the Maclaurin series, i.e.,

$$f(x)\approx p(x)=1+\frac{1}{2!}x+\frac{1}{3!}x^2+\cdots+\frac{1}{8!}x^7.$$

Evaluate it at the same values of $x$ as in part (b).

```python
from math import factorial

def maclaurin(x):
    return sum([(1 / factorial(n)) * (x ** (n - 1)) for n in range(1, 8 + 1)])

maclaurin_algorithm = []
for x in x_values:
    y = maclaurin(x)
    maclaurin_algorithm.append(y)
    print(f"p({x:e}) = {y}")
```

(d) Make a table of the relative difference between the two algorithms as a function of $x$. Which algorithm is more accurate, and why?

```python
def relative_difference(x_tilda, x):
    return abs((x_tilda - x) / x)

for i in range(len(x_values)):
    print(f"Relative difference for x = {x_values[i]:e}: {relative_difference(obvious_algorithm[i], maclaurin_algorithm[i])}")
```

The Maclaurin algorithm is more accurate. The obvious algorithm performs a subtraction between two nearly identical numbers, $e^\epsilon$ and $1$, which has very poor conditioning. The Maclaurin algorithm avoids this. Using an evaluation at $x=10^{-11}$ out to 64 decimal digits as the ground truth, the obvious algorithm gets $7.082$ significant digits, whereas the Maclaurin algorithm gets *at least* $22$; WolframAlpha actually thinks the relative error is 0 and can't take the log of it.

The function

$$x=\cosh(y)=\frac{e^y+e^{-y}}{2}$$

can be inverted to yield a formula for $\text{acosh}(x)$:

$$y=\text{acosh}(x)=-\ln(x-\sqrt{x^2-1}).$$

For the steps below, define $y_i=4i$ and $x_i=\cosh(y_i)$ for $i=1,\dots,4$. Hence $y_i=\text{acosh}(x_i)$.

(a) Find the relative condition number of evaluating $f(x)=\text{acosh}(x)$. (You can use (1.4.4) or look up a formula for $f'$ in a calculus book.) Evaluate $\kappa_f$ at all the $x_i$. (You will find that $\kappa$ is less than 20 at these inputs, so the conditioning is fine.)

$$\kappa_f(x)=|\frac{xf'(x)}{f(x)}|=|\frac{x}{\ln(x-\sqrt{x^2-1})\sqrt{x^2-1}}|$$

$$\begin{align*}
\kappa_f(x_1)=\kappa_f(\frac{e^4+e^{-4}}{2})&=4\\
\kappa_f(x_2)=\kappa_f(\frac{e^8+e^{-8}}{2})&=8\\
\kappa_f(x_3)=\kappa_f(\frac{e^{12}+e^{-12}}{2})&=12\\
\kappa_f(x_4)=\kappa_f(\frac{e^{16}+e^{-16}}{2})&=16
\end{align*}$$

(b) Use (1.4.4) to approximate $f(x_i)$ for all $i$. Compute the relative accuracy of the results. Why are some of the results so inaccurate?

```python
import numpy as np

def cosh(y):
    return (np.exp(y) + np.exp(-1 * y)) / 2

def f(x):
    return -1 * np.log(x - np.sqrt(x ** 2 - 1))

for i in range(1, 4 + 1):
    y = 4 * i
    x = cosh(y)
    f_x = f(x)
    print(f"f({x}) = {f_x}; RE = {abs((f_x - y) / y)}")
```

The results are so inaccurate because $(x_i)^2$, used by the naiive $\text{acosh}$ formula, can get supremely massive; $(x_4)^2$ is something like $2\cdot10^{13}$. Floating points cannot accurately represent supremely massive irrational numbers, so scaling them down with $\text{acosh}$ can show the precision errors seen above.

(c) An alternative formula is

$$y=2\log\left(\sqrt{\frac{x+1}{2}}+\sqrt{\frac{x-1}{2}}\right).$$

Apply (1.4.5) to approximate $f(x_i)$ for all $i$, again computing the relative accuracy of the results.

```python
def f_alternative(x):
    return 2 * np.log(np.sqrt((x + 1) / 2) + np.sqrt((x - 1) / 2))

for i in range(1, 4 + 1):
    y = 4 * i
    x = cosh(y)
    f_x = f_alternative(x)
    print(f"f({x}) = {f_x}; RE = {abs((f_x - y) / y)}")
```
