**Exercise 1.1.2**

Prove that (1.1.5) is equivalent to (1.1.6). This means showing first that (1.1.5) implies (1.1.6), and then separately that (1.1.6) implies (1.1.5).

(1.1.5) states $\frac{|\text{fl}(x)-x|}{|x|}\leq\frac{2^{n-d-1}}{2^n}=\frac{1}{2}\epsilon_M$, and (1.1.6) states $\text{fl}(x)=x(1+\epsilon)$ for some $|\epsilon|\leq\frac{1}{2}\epsilon_M$.

(1.1.5) $\to$ (1.1.6):

$$\begin{align*}
&\frac{|\text{fl}(x)-x|}{|x|}=|\frac{\text{fl}(x)-x}{x}|\leq\frac{1}{2}\epsilon_M\\
\to&\,-\frac{1}{2}\epsilon_M\leq\frac{\text{fl}(x)-x}{x}\leq\frac{1}{2}\epsilon_M\\
\end{align*}$$

Now assume $x>0$.

$$\begin{align*}
\to&\,\text{fl}(x)-x\leq x\frac{1}{2}\epsilon_M\\
\to&\,\text{fl}(x)\leq x(1+\frac{1}{2}\epsilon_M)
\end{align*}$$

Now if we set $\text{fl}(x)=x(1+\epsilon)$ where $\epsilon\leq\frac{1}{2}\epsilon_M$, we get exactly (1.1.6) for positive $\epsilon$.

If we instead assume $x<0$,

$$\begin{align*}
\to&\,-x\frac{1}{2}\epsilon_M\geq\text{fl}(x)-x\\
\to&\,x(1-\frac{1}{2}\epsilon_M)\geq\text{fl}(x)\\
\end{align*}$$

Now if we set $\text{fl}(x)=x(1+\epsilon)$ where $-\epsilon\leq\frac{1}{2}\epsilon_M$, we get exactly (1.1.6) for negative $\epsilon$.

We can combine these into the given $|\epsilon|\leq\frac{1}{2}\epsilon_M$.

(1.1.6) $\to$ (1.1.5):

$$\begin{align*}
&\text{fl}(x)=x(1+\epsilon),|\epsilon|\leq\frac{1}{2}\epsilon_M\\
\to&\,-\frac{1}{2}\epsilon_M\leq\epsilon\leq\frac{1}{2}\epsilon_M\\
\end{align*}$$

Now assume positive $\epsilon$.

$$\to\,\text{fl}(x)\leq x(1+\frac{1}{2}\epsilon_M)$$

We can walk back the algebraic manipulation shown in (1.1.5) $\to$ (1.1.6) to reach $\frac{|\text{fl}(x)-x|}{|x|}\leq\frac{1}{2}\epsilon_M$.

If we instead assume negative $\epsilon$,

$$\to\,x(1-\frac{1}{2}\epsilon_M)\geq\text{fl}(x)$$

We can again walk back the algebraic manipulation to reach the same $\frac{|\text{fl}(x)-x|}{|x|}\leq\frac{1}{2}\epsilon_M$.



As (1.1.5) $\to$ (1.1.6) and (1.1.6) $\to$ (1.1.5), they are equivalent.

(b) $f(x)=\ln x$

(d) $f(x)=e^x$

$$\kappa_f(x)=|\frac{x(e^x)'}{e^x}|=|x|$$

Calculate the relative condition number of each function, and identify all values of $x$ at which $\kappa_f(x)\to\infty$ (including limits as $x\to\pm\infty$).

(a) $f(x)=\tanh x$

$$\kappa_f(x)=|\frac{x(\tanh x)'}{\tanh x}|=|\frac{x\,\text{sech}^2 x}{\tanh x}|=|\frac{x}{\text{sinh}\,x\,\text{cosh}\,x}|$$

```python
import numpy as np
import matplotlib.pyplot as plt

x_axis = np.arange(-10, 10, 0.1)
plt.plot(x_axis, [abs(x / (np.sinh(x) * np.cosh(x))) for x in x_axis])
```

There are no values of $x$ that tend to infinity.

(b) $f(x)=\frac{e^x - 1}{x}$

$$\kappa_f(x)=|\frac{x(x^{-1}(e^x-1))'}{x^{-1}(e^x-1)}|=|\frac{e^x(x-1)+1}{e^x-1}|$$

```python
import numpy as np
import matplotlib.pyplot as plt

x_axis = np.arange(-10, 10, 0.1)
plt.plot(x_axis, [abs((np.exp(x) * (x - 1) + 1) / (np.exp(x) - 1)) for x in x_axis])
```

As $x\to\infty$, $\kappa_f(x)\to\infty$.

```python
import numpy as np
import matplotlib.pyplot as plt

x_axis = np.arange(-10, 10, 0.1)
plt.plot(x_axis, [abs((x - 1) / np.tan(x / 2)) for x in x_axis])
```
