(a) Does Theorem 6.1.1 apply to this problem?

Theorem 6.1.1 does not apply to this problem. $\partial f/\partial u=2u$, which isn't bounded by a constant $L$ since $u$ can be arbitrarily large.

$$\begin{align*}
u'(t)&=(\alpha(1-\alpha t)^{-1})'\\
&=-\alpha(1-\alpha t)^2\cdot(-\alpha)\\
&=\alpha^2(1-\alpha t)^{-2}\\
&=(\alpha/(1-\alpha t))^2\\

\end{align*}$$

Yes, since there is no $t$ term in $u(t)$ that could create a discontinuity.

**Exercise 6.5.1**

Using Function 6.5.2 with an error tolerance of $10^{-8}$, solve $y''+(1+y')^3y=0$ over $0\leq t\leq4\pi$ with the indicated initial conditions. Plot $y(t)$ and $y'(t)$ as functions of $t$ and separately plot the time step size as a function of $t$.

```python
def rk23(du_dt, tspan, u0, tol):
    """
    rk23(du_dt, tspan, u0, tol)

    Apply adaptive embedded RK formula to solve the vector-valued IVP u'=du_dt(u,p,t)
    over the interval tspan with u(tspan[1])=u0, with error tolerance tol.
    Return a vector of times and a vector of solution values/vectors.
    """
    # Initialize for the first time step.
    t = [tspan[0]]
    u = np.array([np.atleast_1d(u0)])
    i = 0
    h = 0.5 * tol ** (1 / 3)
    s1 = du_dt(t[0], u[0])

    # Time stepping.
    while t[i] < tspan[-1]:
        # Detect underflow of the step size.
        if t[i] + h == t[i]:
            print(f"Stepsize too small near t={t[i]}")
            break  # quit time stepping loop

        # New RK stages.
        s2 = du_dt(t[i] + h / 2, u[i] + (h / 2) * s1)
        s3 = du_dt(t[i] + 3 * h / 4, u[i] + (3 * h / 4) * s2)
        unew2 = u[i] + h * (2 * s1 + 3 * s2 + 4 * s3) / 9  # 2rd order solution
        s4 = du_dt(t[i] + h, unew2)
        err = h * (-5 * s1 / 72 + s2 / 12 + s3 / 9 - s4 / 8)  # 2nd/3rd order difference
        E = np.linalg.norm(err, np.inf)  # error estimate
        maxerr = tol * (1 + np.linalg.norm(u[i], np.inf))  # relative/absolute blend

        # Accept the proposed step?
        if E < maxerr:  # yes
            t.append(t[i] + h)
            u = np.vstack([u, unew2])
            i += 1
            s1 = s4  # use FSAL property

        # Adjust step size.
        q = 0.8 * (maxerr / E) ** (1 / 3)  # conservative optimal step factor
        q = min(q, 4)  # limit stepsize growth
        h = min(q * h, tspan[-1] - t[i])  # don't step past the end

    # Convert outputs to arrays
    return np.array(t), u.T
```

(a) $y(0)=0.1,y'(0)=0$

```python
f = lambda t, u: np.array([u[1], -(1+u[1]) ** 3 * u[0]])
t, u = rk23(f, [0, 4 * np.pi], [0.1, 0], 1e-8)

plt.plot(t, u[0, :], label = "y")
plt.plot(t, u[1, :], label = "y'")
plt.xlabel("t")
plt.legend()
plt.show()
```
